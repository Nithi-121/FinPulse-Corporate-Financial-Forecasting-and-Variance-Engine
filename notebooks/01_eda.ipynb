# FinPulse exploratory analysis

This notebook records descriptive hypotheses before forecasting. Results are exploratory, not causal; several series contain missing quarters and issuer-specific metric gaps documented in `DATA_QUALITY.md`. The script saves a hypothesis matrix, ADF diagnostics, and six charts. ADF tests on short quarterly histories are diagnostics only.

```python
from pathlib import Path
import runpy
from IPython.display import display, Image

candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next(path for path in candidates if (path / 'data' / 'finpulse.duckdb').exists())
runpy.run_path(str(ROOT / 'src' / 'eda.py'), run_name='__main__')
```

```python
matrix = __import__('pandas').read_csv(ROOT / 'data' / 'processed' / 'hypothesis_matrix.csv')
display(matrix)

for chart in sorted((ROOT / 'dashboard' / 'eda').glob('*.png')):
    display(Image(filename=str(chart)))
```
