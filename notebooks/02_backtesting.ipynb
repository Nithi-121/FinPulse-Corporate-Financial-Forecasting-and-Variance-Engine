# FinPulse rolling-origin revenue backtest

This walkthrough runs the reproducible one-quarter-ahead benchmark. It keeps only uninterrupted reported quarters; model rankings are exploratory because eligible histories are short and latest-restated. See `BACKTEST_FINDINGS.md` for the protocol and limitations.

```python
from pathlib import Path
import runpy
from IPython.display import display, Image
import pandas as pd

candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next(path for path in candidates if (path / 'data' / 'finpulse.duckdb').exists())
runpy.run_path(str(ROOT / 'src' / 'backtest.py'), run_name='__main__')
```

```python
metrics = pd.read_csv(ROOT / 'data' / 'processed' / 'backtest_metrics.csv')
eligibility = pd.read_csv(ROOT / 'data' / 'processed' / 'backtest_eligibility.csv')
display(eligibility)
display(metrics.sort_values(['company', 'mae']))
display(Image(filename=str(ROOT / 'dashboard' / 'backtest' / 'mase_by_company.png')))
```
